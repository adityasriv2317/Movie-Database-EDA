```python
genre_stats=df.groupby(['Genre','ReleaseYear']).agg({
                      "Global_BoxOfficeUSD":'mean',
                      "BudgetUSD":'mean',
                      'IMDbRating':'mean',
                      'RottenTomatoesScore':'mean'
                      }).reset_index()
```

```python
df.columns
```

```python
genre_stats
```

```python
plt.figure(figsize=(15,6))
sns.lineplot(data=genre_stats,x="ReleaseYear",y="Global_BoxOfficeUSD", hue="Genre")
plt.title("Average IT Box Office")
plt.show()
```

```python
plt.figure(figsize=(15,6))
sns.lineplot(data=genre_stats.query("Genre == 'Action'"),x="ReleaseYear",y="Global_BoxOfficeUSD", hue="Genre")
plt.title("Average Action Box Office")
plt.show()
```
